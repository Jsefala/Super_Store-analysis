```python
#Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

```python
#remove unnecesary space
text_columns = df.select_dtypes(include="string").columns

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
#convert the date formats
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"], errors="coerce")
```

```python
#view data
df[["Order_Date", "Ship_Date"]].head()
```

```python
#convert numeric columns
numeric_columns = [
    "Row_ID",
    "Postal_Code",
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
#check for missing values
df.isnull().sum()
```

```python
#calculate total
df.isnull().sum().sum()
```

```python
#check for duplicates
df.duplicated().sum()
```

```python
#remove duplicates
df = df.drop_duplicates()
```

```python
#check invalid quantities
df[df["Quantity"] <= 0]
```

```python
#check invalid quantities
df = df[df["Quantity"] > 0]
```

```python
#check invalid discounts
df[(df["Discount"] < 0) | (df["Discount"] > 1)]
```

```python
#check invalid discounts
df["Discount"] = df["Discount"].clip(0, 1)
```

```python
#calculate shipping days
df["Shipping_Days"] = (
    df["Ship_Date"] - df["Order_Date"]
).dt.days
```

```python
df[["Order_Date", "Ship_Date", "Shipping_Days"]].head()
```

```python
#create order year
df["Order_Year"] = df["Order_Date"].dt.year
```

```python
#create order month
df["Order_Month"] = df["Order_Date"].dt.month
```

```python
#create order month
df["Order_Month_Name"] = df["Order_Date"].dt.strftime("%B")
```

```python
#create order quarter
df["Order_Quarter"] = "Q" + df["Order_Date"].dt.quarter.astype(str)
```

```python
#calculate order sales
df["Gross_Sales"] = np.where(
    df["Discount"] < 1,
    df["Sales"] / (1 - df["Discount"]),
    np.nan
)
```

```python
#calculate discount amount
df["Discount_Amount"] = (
    df["Gross_Sales"] - df["Sales"]
)
```

```python
#calculate sale per unit
df["Sales_Per_Unit"] = (
    df["Sales"] / df["Quantity"]
)
```

```python
#calculate profit margin
df["Profit_Margin"] = np.where(
    df["Sales"] != 0,
    df["Profit"] / df["Sales"],
    np.nan
)
```

```python
#view profit percentage
df["Profit_Margin"] = df["Profit_Margin"] * 100
```

```python
#create profit status
df["Profit_Status"] = np.where(
    df["Profit"] >= 0,
    "Profitable",
    "Loss"
)
```

```python
#rounf calculated values
df["Gross_Sales"] = df["Gross_Sales"].round(2)

df["Discount_Amount"] = df["Discount_Amount"].round(2)

df["Sales_Per_Unit"] = df["Sales_Per_Unit"].round(2)

df["Profit_Margin"] = df["Profit_Margin"].round(2)
```
